## Load the dataset and inspect it

```python
n_total = len(ds["train"])
n_ans = sum(len(ex["answer"]) > 0 for ex in ds["train"])
print(f"train rows: {n_total}, answerable: {n_ans}")
```

```python
n_total = len(ds["train"])
n_ans = sum(not ex["is_impossible"] for ex in ds["train"])
print(f"train rows: {n_total}, answerable: {n_ans}")
```
